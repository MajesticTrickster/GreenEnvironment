```python
# Use the GreenEx_Py module
import geopandas as gpd
from GreenExp import availability, accessibility, visibility
```

```python
# Path to input data
filepath = "TestData/"

# Filename for points of interest
poi_file = filepath+"AMS_example_data.gpkg"
# The poi_file object will serve as input for the functions
```

```python
# The example data used for this notebook looks as follows
example_data = gpd.read_file(filepath + "AMS_example_data.gpkg")
example_data
```

# Availability

```python
# NDVI
availability.get_mean_NDVI(point_of_interest=poi_file,
                           buffer_type="network",
                           travel_speed=5,
                           trip_time=15,
                           network_type="walk",
                           plot_aoi=True,
                           save_ndvi=False,
                           write_to_file=False)
```

```python
# Land cover class percentages
availability.get_landcover_percentages(point_of_interest=poi_file,
                                       buffer_dist=500,
                                       buffer_type="euclidean",
                                       plot_aoi=True,
                                       save_lulc=False,
                                       write_to_file=False)
```

```python
# Use different poi_file due to availability of tree canopy data
single_poi_file = filepath+"AMS_canopy_example.gpkg"

# Canopy coverage
availability.get_canopy_percentage(point_of_interest=single_poi_file,
                                   canopy_vector_file=filepath+"AMS_canopy_segment.gpkg",
                                   buffer_type="network",
                                   travel_speed=15,
                                   trip_time=10,
                                   network_type="bike",
                                   plot_aoi=True,
                                   write_to_file=False)
```

```python
# Park area coverage
availability.get_greenspace_percentage(point_of_interest=poi_file,
                                       buffer_type="network",
                                       travel_speed=5,
                                       trip_time=15,
                                       network_type="walk",
                                       write_to_file=False)
```

# Accessibility

```python
# Shortest distance to park area
accessibility.get_shortest_distance_greenspace(point_of_interest=poi_file,
                                               target_dist=300,
                                               distance_type='euclidean',
                                               destination='centroids',
                                               min_greenspace_area=400,
                                               write_to_file=False)
```

# Visibility

```python
poi, sampled_points_df = visibility.get_viewshed_GVI(point_of_interest=poi_file,
                                                     greendata_raster_file=filepath+"AMS_trees_binary_crs.tif",
                                                     dtm_raster_file=filepath+"AMS_DTM_crs.tif",
                                                     dsm_raster_file=filepath+"AMS_DSM_crs.tif",
                                                     buffer_dist=100,
                                                     viewing_dist=250,
                                                     sample_dist=50,
                                                     observer_height=1.7,
                                                     write_to_file=False)
```
